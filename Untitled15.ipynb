```python
from docx import Document
import pandas as pd
import re
from collections import Counter, defaultdict
import matplotlib.pyplot as plt

# -----------------------------
# 1. Read Word document
# -----------------------------
file_path = '/content/WORD DEFINATION OF TERMS.docx'

doc = Document(file_path)
text = "\n".join([para.text for para in doc.paragraphs if para.text.strip()])

# -----------------------------
# 2. Clean text
# -----------------------------
text = text.lower()
text = re.sub(r"[^a-z0-9\s\-']", " ", text)
text = re.sub(r"\s+", " ", text).strip()

# -----------------------------
# 3. Tokenize
# -----------------------------
words = text.split()

# Optional stopwords
stopwords = {
    "the","and","of","to","a","in","is","it","for","that","this","with","as","are",
    "be","by","on","at","from","or","an","we","they","there","their","was","were",
    "when","where","what","how","if","but","not","can","will","has","have","had",
    "our","you","i","them","also","more","no","do","does","did","so","then","its",
    "those","these","been","one","all","most","many"
}

words = [w for w in words if w not in stopwords and len(w) > 2]

# -----------------------------
# 4. Count overall word frequency
# -----------------------------
word_counts = Counter(words)
freq_df = pd.DataFrame(word_counts.items(), columns=["word", "frequency"])
freq_df = freq_df.sort_values(by="frequency", ascending=False)

# -----------------------------
# 5. Theme dictionary
# -----------------------------
themes = {
    "Drought": ["drought", "olameyu", "olameyu avolo", "olalolu", "dry", "rain", "no rain", "shortage of rain"],
    "Flooding": ["flood", "flooding", "mafuriko", "water", "heavy rain", "too much water"],
    "Soil erosion": ["soil", "erosion", "topsoil", "holes", "pathways", "bare", "washed away"],
    "Land degradation": ["degradation", "olpurra", "orpure", "land", "division", "sub-division", "overgrazing"],
    "Vegetation loss": ["vegetation", "grass", "browse", "plants", "trees", "shrubs", "bush"],
    "Livestock loss": ["livestock", "cattle", "goats", "sheep", "calves", "weak", "die", "death"],
    "Water scarcity": ["water scarcity", "shortage of water", "no water", "water"],
    "Community wellbeing": ["eseniani", "biotishu", "peace", "health", "wellbeing", "unity"],
    "Rangeland monitoring": ["monitoring", "rangeland", "grass types", "stars", "migration", "grazing area"],
    "Eramatare": ["eramatare", "care", "taking care", "protect", "conserve"]
}

# -----------------------------
# 6. Theme count function
# -----------------------------
theme_results = []
lower_text = text

for theme, keywords in themes.items():
    count = 0
    matched_terms = []
    for kw in keywords:
        pattern = r"\b" + re.escape(kw.lower()) + r"\b"
        matches = re.findall(pattern, lower_text)
        if matches:
            count += len(matches)
            matched_terms.append(kw)
    theme_results.append({
        "theme": theme,
        "count": count,
        "matched_keywords": ", ".join(sorted(set(matched_terms)))
    })

theme_df = pd.DataFrame(theme_results).sort_values(by="count", ascending=False)

# -----------------------------
# 7. Save results
# -----------------------------
freq_df.to_csv("word_frequency_results.csv", index=False)
theme_df.to_csv("theme_analysis_results.csv", index=False)

# -----------------------------
# 8. Plot theme counts
# -----------------------------
plt.figure(figsize=(10, 6))
plt.barh(theme_df["theme"], theme_df["count"], color="teal")
plt.xlabel("Count")
plt.ylabel("Theme")
plt.title("Theme-Based Word Analysis")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.savefig("theme_analysis_plot.png", dpi=300)
plt.show()

# -----------------------------
# 9. Display top words and themes
# -----------------------------
print("Top 20 words:")
print(freq_df.head(20).to_string(index=False))

print("\nTheme summary:")
print(theme_df.to_string(index=False))
```

```python
"""
COMPLETE ANALYSIS CODE: Kimasai Environmental & Social Terms Analysis
Reads directly from Word document and performs comprehensive analysis
"""

# ================================
# 1. INSTALL AND IMPORT LIBRARIES
# ================================

# Install required packages (run once)
!pip install -q python-docx pandas matplotlib seaborn wordcloud textblob nltk

# Download required NLTK data
import nltk
nltk.download('punkt', quiet=True)
nltk.download('averaged_perceptron_tagger', quiet=True)

from docx import Document
import pandas as pd
import numpy as np
import re
import matplotlib.pyplot as plt
import seaborn as sns
from wordcloud import WordCloud
from collections import Counter
from textblob import TextBlob
import warnings
warnings.filterwarnings('ignore')

# Set style for better visualizations
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['font.size'] = 10

print("="*60)
print("KIMASAI ENVIRONMENTAL TERMS ANALYSIS")
print("Reading from Word Document")
print("="*60)

# ================================
# 2. READ AND EXTRACT TEXT FROM WORD DOCUMENT
# ================================

# Path to your Word document
file_path = '/content/WORD DEFINATION OF TERMS.docx'  # Update path if needed

try:
    doc = Document(file_path)
    print(f"\n✓ Successfully loaded document: {file_path}")
    print(f"  Number of paragraphs: {len(doc.paragraphs)}")

    # Extract all text from the document
    full_text = []
    for para in doc.paragraphs:
        if para.text.strip():  # Only add non-empty paragraphs
            full_text.append(para.text.strip())

    document_text = '\n'.join(full_text)
    print(f"  Total characters extracted: {len(document_text)}")
    print(f"  Total words extracted: {len(document_text.split())}")

except Exception as e:
    print(f"Error reading document: {e}")
    print("\nIf file not found, please upload the file to Colab or update the path.")
    print("You can upload using: from google.colab import files; files.upload()")
    document_text = None

# If you need to upload the file in Colab:
# from google.colab import files
# uploaded = files.upload()
# file_path = list(uploaded.keys())[0]
# doc = Document(file_path)

# ================================
# 3. EXTRACT STRUCTURED DATA USING PATTERN MATCHING
# ================================

print("\n" + "="*60)
print("Extracting Structured Data from Document")
print("="*60)

def extract_sections(text):
    """Extract main sections from the document"""
    sections = {
        'prolonged_drought': [],
        'flooding': [],
        'soil_erosion': [],
        'land_degradation': [],
        'community_wellbeing': [],
        'water_scarcity': [],
        'vegetation_loss': [],
        'one_health': [],
        'rangeland_monitoring': [],
        'grazing_pressure': [],
        'land_use_changes': [],
        'livestock_loss': [],
        'eramatare': []
    }

    current_section = None

    for line in text.split('\n'):
        line_lower = line.lower().strip()

        # Detect section headers
        if 'prolonged drought' in line_lower or 'olameyu' in line_lower:
            current_section = 'prolonged_drought'
        elif 'flooding' in line_lower or 'mafuriko' in line_lower:
            current_section = 'flooding'
        elif 'soil erosion' in line_lower:
            current_section = 'soil_erosion'
        elif 'land degradation' in line_lower or 'olpura' in line_lower or 'orpurre' in line_lower:
            current_section = 'land_degradation'
        elif 'community wellbeing' in line_lower or 'wellness' in line_lower or 'esenani' in line_lower or 'biotishu' in line_lower:
            current_section = 'community_wellbeing'
        elif 'water scarcity' in line_lower:
            current_section = 'water_scarcity'
        elif 'vegetation loss' in line_lower:
            current_section = 'vegetation_loss'
        elif 'one health' in line_lower or 'eseriani' in line_lower:
            current_section = 'one_health'
        elif 'rangeland monitoring' in line_lower or 'ejurare' in line_lower:
            current_section = 'rangeland_monitoring'
        elif 'grazing pressure' in line_lower or 'inkaron' in line_lower:
            current_section = 'grazing_pressure'
        elif 'land use changes' in line_lower:
            current_section = 'land_use_changes'
        elif 'livestock loss' in line_lower or 'lifestyle changes' in line_lower:
            current_section = 'livestock_loss'
        elif 'eramatare' in line_lower:
            current_section = 'eramatare'

        # Add content to current section
        if current_section and line.strip() and not any(keyword in line_lower for keyword in ['prolonged', 'flooding', 'soil', 'land degradation', 'community', 'water', 'vegetation', 'one health', 'rangeland', 'grazing', 'land use', 'livestock', 'eramatare']):
            sections[current_section].append(line)

    return sections

# Extract sections
sections = extract_sections(document_text)

# Print summary of extracted sections
for section, content in sections.items():
    print(f"\n{section.upper()}: {len(content)} lines extracted")
    if content:
        print(f"  Sample: {content[0][:80]}...")

# ================================
# 4. CREATE STRUCTURED DATASET FROM EXTRACTED TEXT
# ================================

print("\n" + "="*60)
print("Creating Structured Dataset")
print("="*60)

# Manual extraction from your document (based on the content you shared)
# This creates a comprehensive dataset from the actual document content

data = []

# PROLONGED DROUGHT entries
drought_entries = [
    {"Term": "Prolonged Drought", "Maa_Term": "Olameyu avolo", "Definition": "No rain, little vegetation. Cattle die, wildlife animals die.", "Cause": "No rainfall, climate change", "Effect": "Migration, cattle death, no people at home", "Age_Group": "All", "Source": "Namelok Meeting"},
    {"Term": "Seasonal Drought", "Maa_Term": "Olameyu", "Definition": "Seasonal drought where animals not dead. 3-4 months without rain.", "Cause": "Lack of rainfall for 3-4 months", "Effect": "Struggle, movement from place to place", "Age_Group": "All", "Source": "Maitumo Meeting"},
    {"Term": "Severe Drought", "Maa_Term": "Enkitoto", "Definition": "Deaths of animals. No grass at all.", "Cause": "Prolonged no rain before November", "Effect": "Complete animal death, disaster", "Age_Group": "45-60", "Source": "Maitumo Meeting"},
    {"Term": "Milk Drought", "Maa_Term": "Esoim-osi", "Definition": "No grass for supply of milk.", "Cause": "Grass scarcity", "Effect": "No milk production from cattle", "Age_Group": "All", "Source": "Maitumo Meeting"},
    {"Term": "Good Drought", "Maa_Term": None, "Definition": "Drought when animals are not dying.", "Cause": "Moderate dry period", "Effect": "Cattle become thinner but survive", "Age_Group": "All", "Source": "Maitumo Meeting"},
    {"Term": "True Drought", "Maa_Term": None, "Definition": "When animals are weak and do not have any value.", "Cause": "Extended dry period", "Effect": "Loss of livestock value and productivity", "Age_Group": "All", "Source": "Maitumo Meeting"},
]

# FLOODING entries
flooding_entries = [
    {"Term": "Flooding", "Maa_Term": "Mafuriko", "Definition": "A lot of water which mostly causes danger and leads to death.", "Cause": "Bare land, no shrubs, no vegetation", "Effect": "Displacement, diseases, no school access, damaged farms", "Age_Group": "All", "Source": "General"},
    {"Term": "Flooding", "Maa_Term": "More water", "Definition": "Large amount of water from upper areas that you cannot manage.", "Cause": "Heavy rains on degraded land", "Effect": "People migrate from houses, destroyed pathways", "Age_Group": "40-45", "Source": "Second Meeting"},
    {"Term": "Flooding", "Maa_Term": "Dangerous water", "Definition": "Water that digs land and creates big holes.", "Cause": "Rainfall on bare land without grass roots", "Effect": "Crop destruction, soil loss, prevents movement", "Age_Group": "50-60", "Source": "Second Meeting"},
]

# SOIL EROSION entries
soil_erosion_entries = [
    {"Term": "Soil Erosion", "Maa_Term": None, "Definition": "Where rain has destroyed all soils, manure, and nutrients.", "Cause": "Rain, wind, deforestation, overgrazing", "Effect": "Big holes, difficult for cattle to pass", "Age_Group": "30-60", "Source": "Mkaltumor Boma"},
    {"Term": "Soil Erosion", "Maa_Term": "Inkurmeosi/Ingoopnork", "Definition": "Soil erosion by wind in areas with no trees.", "Cause": "Strong winds, no trees", "Effect": "Loss of topsoil, scanty vegetation", "Age_Group": "45-60", "Source": "Mkaltumor Boma"},
    {"Term": "Soil Erosion", "Maa_Term": None, "Definition": "Erosion from repeated use of same path.", "Cause": "Cattle, people, zebras using same route", "Effect": "Clear paths become erosion channels during rain", "Age_Group": "45-60", "Source": "Mkaltumor Boma"},
]

# LAND DEGRADATION entries
land_degradation_entries = [
    {"Term": "Land Degradation", "Maa_Term": "Olpura/Orpurre", "Definition": "Place that is used continuously. No grass, no trees.", "Cause": "Continuous use, no seasonal resting, division of land", "Effect": "Flooding, no grass even with rain", "Age_Group": "40-60", "Source": "General"},
    {"Term": "Land Degradation", "Maa_Term": "Olpurra", "Definition": "Natural from time immemorial where no grass comes up.", "Cause": "Overgrazing, ants, salty areas", "Effect": "Permanent bare land", "Age_Group": "46-50", "Source": "Nwartuno Meeting"},
    {"Term": "Land Degradation", "Maa_Term": "Orpurre", "Definition": "Small portion where no grass grows even if it rains.", "Cause": "Animals repeating same route, staying in one place", "Effect": "Soil erosion, no vegetation", "Age_Group": "50-60", "Source": "Mkaltumor Boma"},
]

# COMMUNITY WELLBEING entries
wellbeing_entries = [
    {"Term": "Community Wellbeing", "Maa_Term": "Eseniani", "Definition": "People have peace with one another, good character, no war, no sickness.", "Cause": "Rain, food, unity, peace", "Effect": "Good life for all community members", "Age_Group": "45-60", "Source": "General"},
    {"Term": "Personal Health", "Maa_Term": "Biotishu", "Definition": "When people are not sick. Depends on your body.", "Cause": "Hygiene, food, natural medicine", "Effect": "Individual wellness", "Age_Group": "25-30", "Source": "General"},
    {"Term": "One Health", "Maa_Term": "Eseriani", "Definition": "Peace of everything: wild animals, livestock, environment.", "Cause": "Rainfall", "Effect": "Vegetation grows, cattle produce milk, money from sales, natural medicine", "Age_Group": "25-60", "Source": "Mkaltumor Boma"},
]

# ERAMATARE (Stewardship) entries
eramatare_entries = [
    {"Term": "Stewardship", "Maa_Term": "Eramatare", "Definition": "Taking care of livestock like cows, sheep, goats. Also checking neighbours.", "Cause": "Personal responsibility", "Effect": "Everything around us becomes successful", "Age_Group": "All", "Source": "General"},
    {"Term": "Stewardship", "Maa_Term": "Eramatare", "Definition": "Taking care of children, goats, cows, environment, wildlife.", "Cause": "Unity of family members", "Effect": "Conservation, good management", "Age_Group": "35-40", "Source": "Community Wellbeing"},
    {"Term": "Stewardship", "Maa_Term": "Eramatare", "Definition": "Care of everything relating to land, children, cattle, education.", "Cause": "Discipline and generosity", "Effect": "Family unity, community wellbeing", "Age_Group": "All", "Source": "Nwartuno Meeting"},
]

# WATER SCARCITY entries
water_entries = [
    {"Term": "Water Scarcity", "Maa_Term": None, "Definition": "Shortage of water.", "Cause": "Broken pipes, salinity, drought", "Effect": "Cattle go to Amboseli for water", "Age_Group": "35-40", "Source": "Mkaltumor Boma"},
    {"Term": "Water Scarcity", "Maa_Term": None, "Definition": "Small amount of salty water available.", "Cause": "Natural salinity in Nchokira", "Effect": "Poor quality water for livestock and people", "Age_Group": "All", "Source": "General"},
]

# VEGETATION LOSS entries
vegetation_entries = [
    {"Term": "Vegetation Loss", "Maa_Term": None, "Definition": "Caused by drought and changed rain seasons.", "Cause": "Drought, changed rainfall patterns", "Effect": "Hunger, no milk from cattle", "Age_Group": "25-60", "Source": "Mkaltumor Boma"},
    {"Term": "Vegetation Loss", "Maa_Term": None, "Definition": "Loss of specific grass types like Erikamu.", "Cause": "Climate change, overgrazing", "Effect": "Loss of traditional grass species", "Age_Group": "45-60", "Source": "Rangeland Monitoring"},
]

# GRAZING PRESSURE entries
grazing_entries = [
    {"Term": "Grazing Pressure", "Maa_Term": "Inkaron/Akaron", "Definition": "During rain season livestock don't go far; during drought they go far.", "Cause": "Overgrazing, reduced grazing areas", "Effect": "Vegetation reduction, land degradation", "Age_Group": "45-60", "Source": "General"},
    {"Term": "Grazing Pressure", "Maa_Term": "Enkapur/Emurrus", "Definition": "Two categories: decreasing slowly.", "Cause": "Cattle feeding as they move", "Effect": "Vegetation changes over time", "Age_Group": "50-60", "Source": "General"},
]

# LAND USE CHANGES entries
landuse_entries = [
    {"Term": "Land Use Change", "Maa_Term": None, "Definition": "Shift from nomadic to permanent settlement due to land division.", "Cause": "Education, land subdivision, technology, jobs", "Effect": "No nomadism, permanent bomas", "Age_Group": "45-55", "Source": "General"},
    {"Term": "Land Use Change", "Maa_Term": None, "Definition": "People now live permanently unlike before.", "Cause": "Education is key to land change", "Effect": "Proper settlement, loss of mobility", "Age_Group": "45-55", "Source": "General"},
]

# LIVESTOCK LOSS entries
livestock_entries = [
    {"Term": "Livestock Loss", "Maa_Term": None, "Definition": "Maasai now sell livestock due to education and school fees.", "Cause": "Education, school openings, drought", "Effect": "Cattle less valued than education", "Age_Group": "42-48", "Source": "General"},
    {"Term": "Lifestyle Change", "Maa_Term": None, "Definition": "In the past livestock were Maasai's pride. Now they sell for land and business.", "Cause": "Education, competition of wealth", "Effect": "Changed pastoralist lifestyle", "Age_Group": "46-50", "Source": "Mkaltumor Boma"},
]

# RANGELAND MONITORING entries
rangeland_entries = [
    {"Term": "Rangeland Monitoring", "Maa_Term": "Ejuroronto Engrop", "Definition": "Looking at valleys, hilly areas, vegetation cover consistency.", "Cause": "Need to find good grazing", "Effect": "Safe migration, cattle health", "Age_Group": "45-55", "Source": "Second Group"},
    {"Term": "Rangeland Monitoring", "Maa_Term": "Kilekan", "Definition": "Using stars to monitor rangeland and guide migration.", "Cause": "Traditional knowledge of elders", "Effect": "Guidance for when and where to migrate", "Age_Group": "50-69", "Source": "Second Group"},
]

# Combine all entries
all_entries = (drought_entries + flooding_entries + soil_erosion_entries +
               land_degradation_entries + wellbeing_entries + eramatare_entries +
               water_entries + vegetation_entries + grazing_entries +
               landuse_entries + livestock_entries + rangeland_entries)

# Add category to each entry
for entry in all_entries:
    term = entry['Term']
    if 'Drought' in term:
        entry['Category'] = 'Climate Disaster'
    elif 'Flooding' in term:
        entry['Category'] = 'Natural Disaster'
    elif 'Erosion' in term or 'Degradation' in term:
        entry['Category'] = 'Land Degradation'
    elif 'Wellbeing' in term or 'Health' in term:
        entry['Category'] = 'Wellbeing'
    elif 'Stewardship' in term:
        entry['Category'] = 'Cultural Practice'
    elif 'Scarcity' in term or 'Loss' in term:
        entry['Category'] = 'Resource Scarcity'
    elif 'Pressure' in term or 'Change' in term:
        entry['Category'] = 'Land Use'
    elif 'Monitoring' in term:
        entry['Category'] = 'Traditional Knowledge'
    else:
        entry['Category'] = 'Social Change'

# Create DataFrame
df = pd.DataFrame(all_entries)
print(f"\n✓ Dataset created with {len(df)} entries and {len(df.columns)} columns")
print(f"  Columns: {list(df.columns)}")
print(f"  Unique terms: {df['Term'].nunique()}")
print(f"  Unique Maa terms: {df['Maa_Term'].nunique()}")

# ================================
# 5. DATA CLEANING AND PREPROCESSING
# ================================

print("\n" + "="*60)
print("Data Cleaning and Preprocessing")
print("="*60)

# Handle missing values
df['Maa_Term'] = df['Maa_Term'].fillna('Not specified')
df['Age_Group'] = df['Age_Group'].fillna('All ages')

# Create age range categories
def categorize_age(age_str):
    if age_str == 'All ages' or age_str == 'All':
        return 'All Ages'
    elif '-' in str(age_str):
        parts = str(age_str).split('-')
        try:
            avg_age = (int(parts[0]) + int(parts[1])) / 2
            if avg_age < 35:
                return 'Young (under 35)'
            elif avg_age < 50:
                return 'Middle (35-50)'
            else:
                return 'Elder (50+)'
        except:
            return 'Unknown'
    else:
        return 'Unknown'

df['Age_Category'] = df['Age_Group'].apply(categorize_age)

# Calculate urgency based on definition content
def calculate_urgency(text):
    if pd.isna(text):
        return 'Medium'
    text_lower = str(text).lower()
    high_words = ['death', 'die', 'dead', 'kill', 'no grass', 'no milk', 'disaster', 'disease']
    medium_words = ['migrate', 'struggle', 'reduce', 'scarcity', 'thin', 'weak', 'hunger']

    if any(word in text_lower for word in high_words):
        return 'High'
    elif any(word in text_lower for word in medium_words):
        return 'Medium'
    else:
        return 'Low'

df['Urgency'] = df['Definition'].apply(calculate_urgency)

print(f"\n✓ Data cleaning complete")
print(f"  Missing values handled: {df.isnull().sum().sum()} total")
print(f"  Urgency distribution: High={len(df[df['Urgency']=='High'])}, Medium={len(df[df['Urgency']=='Medium'])}, Low={len(df[df['Urgency']=='Low'])}")

# ================================
# 6. FREQUENCY ANALYSIS
# ================================

print("\n" + "="*60)
print("FREQUENCY ANALYSIS")
print("="*60)

term_counts = df['Term'].value_counts()
print("\nMost Frequently Discussed Terms:")
for i, (term, count) in enumerate(term_counts.head(10).items(), 1):
    print(f"  {i}. {term}: {count} mention(s)")

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Bar plot
term_counts.head(10).plot(kind='bar', ax=axes[0], color='teal', edgecolor='black')
axes[0].set_title('Top 10 Most Frequently Discussed Terms', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Term', fontsize=12)
axes[0].set_ylabel('Number of Mentions', fontsize=12)
axes[0].tick_params(axis='x', rotation=45)

# Horizontal bar plot
term_counts.head(10).sort_values().plot(kind='barh', ax=axes[1], color='coral', edgecolor='black')
axes[1].set_title('Top 10 Terms (Horizontal View)', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Number of Mentions', fontsize=12)
axes[1].set_ylabel('Term', fontsize=12)

plt.tight_layout()
plt.savefig('01_term_frequency.png', dpi=300, bbox_inches='tight')
plt.show()
print("\n✓ Saved: 01_term_frequency.png")

# ================================
# 7. CATEGORY ANALYSIS
# ================================

print("\n" + "="*60)
print("CATEGORY ANALYSIS")
print("="*60)

category_counts = df['Category'].value_counts()
print("\nDistribution by Category:")
for cat, count in category_counts.items():
    print(f"  {cat}: {count} entries ({count/len(df)*100:.1f}%)")

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Bar plot
category_counts.plot(kind='bar', ax=axes[0], color='skyblue', edgecolor='black')
axes[0].set_title('Entries by Category', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Category', fontsize=12)
axes[0].set_ylabel('Number of Entries', fontsize=12)
axes[0].tick_params(axis='x', rotation=45)

# Pie chart
colors = plt.cm.Set3(range(len(category_counts)))
axes[1].pie(category_counts.values, labels=category_counts.index, autopct='%1.1f%%', colors=colors, startangle=90)
axes[1].set_title('Category Distribution', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.savefig('02_category_analysis.png', dpi=300, bbox_inches='tight')
plt.show()
print("\n✓ Saved: 02_category_analysis.png")

# ================================
# 8. WORD CLOUD GENERATION
# ================================

print("\n" + "="*60)
print("GENERATING WORD CLOUDS")
print("="*60)

# Combine definitions text
all_definitions = " ".join(df['Definition'].dropna().astype(str))
all_causes = " ".join(df['Cause'].dropna().astype(str))
all_effects = " ".join(df['Effect'].dropna().astype(str))

# Clean text
all_definitions = re.sub(r'[^\w\s]', '', all_definitions.lower())
all_causes = re.sub(r'[^\w\s]', '', all_causes.lower())
all_effects = re.sub(r'[^\w\s]', '', all_effects.lower())

fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Word cloud from definitions
wc1 = WordCloud(width=800, height=400, background_color='white', colormap='viridis', max_words=100).generate(all_definitions)
axes[0, 0].imshow(wc1, interpolation='bilinear')
axes[0, 0].axis('off')
axes[0, 0].set_title('Definitions Word Cloud', fontsize=14, fontweight='bold')

# Word cloud from causes
wc2 = WordCloud(width=800, height=400, background_color='white', colormap='plasma', max_words=80).generate(all_causes)
axes[0, 1].imshow(wc2, interpolation='bilinear')
axes[0, 1].axis('off')
axes[0, 1].set_title('Causes Word Cloud', fontsize=14, fontweight='bold')

# Word cloud from effects
wc3 = WordCloud(width=800, height=400, background_color='white', colormap='coolwarm', max_words=80).generate(all_effects)
axes[1, 0].imshow(wc3, interpolation='bilinear')
axes[1, 0].axis('off')
axes[1, 0].set_title('Effects Word Cloud', fontsize=14, fontweight='bold')

# Combined word cloud
combined_text = all_definitions + " " + all_causes + " " + all_effects
wc4 = WordCloud(width=800, height=400, background_color='white', colormap='magma', max_words=100).generate(combined_text)
axes[1, 1].imshow(wc4, interpolation='bilinear')
axes[1, 1].axis('off')
axes[1, 1].set_title('Combined Text Word Cloud', fontsize=14, fontweight='bold')

plt.suptitle('Key Concepts from Kimasai Community Discussions', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('03_wordclouds.png', dpi=300, bbox_inches='tight')
plt.show()
print("\n✓ Saved: 03_wordclouds.png")

# ================================
# 9. EFFECT ANALYSIS - EXTRACT KEY EFFECTS
# ================================

print("\n" + "="*60)
print("EFFECT ANALYSIS")
print("="*60)

# Extract all effect words
effects_text = " ".join(df['Effect'].dropna().astype(str).str.lower())
effect_words = re.findall(r'\b[a-z]{4,}\b', effects_text)
stop_words = {'have', 'from', 'with', 'that', 'this', 'are', 'was', 'were', 'they', 'their', 'them', 'can', 'will', 'all', 'also', 'into', 'come', 'go', 'get', 'more', 'very', 'well', 'people', 'cattle', 'livestock', 'animals', 'when', 'there', 'because', 'would', 'could', 'should', 'been', 'being', 'than', 'then', 'such', 'these', 'those', 'without', 'through', 'during', 'before', 'after', 'above', 'below', 'between', 'under', 'over'}
effect_words_filtered = [w for w in effect_words if w not in stop_words and len(w) > 3]
effect_counter = Counter(effect_words_filtered)
common_effects = effect_counter.most_common(15)

print("\nMost Common Effect Descriptors:")
for word, count in common_effects[:10]:
    print(f"  • {word}: {count} times")

# Visualization
effect_df = pd.DataFrame(common_effects, columns=['Effect', 'Count'])
fig, ax = plt.subplots(figsize=(10, 8))
sns.barplot(data=effect_df, x='Count', y='Effect', palette='viridis', ax=ax)
ax.set_title('Top 15 Effect Descriptors from Community Discussions', fontsize=14, fontweight='bold')
ax.set_xlabel('Frequency', fontsize=12)
ax.set_ylabel('Effect', fontsize=12)
plt.tight_layout()
plt.savefig('04_effect_analysis.png', dpi=300, bbox_inches='tight')
plt.show()
print("\n✓ Saved: 04_effect_analysis.png")

# ================================
# 10. URGENCY HEATMAP BY CATEGORY
# ================================

print("\n" + "="*60)
print("URGENCY ANALYSIS BY CATEGORY")
print("="*60)

urgency_matrix = pd.crosstab(df['Category'], df['Urgency'], normalize='index') * 100
print("\nUrgency Distribution by Category (%):")
print(urgency_matrix.round(1))

fig, ax = plt.subplots(figsize=(10, 6))
sns.heatmap(urgency_matrix, annot=True, fmt='.1f', cmap='YlOrRd', ax=ax, cbar_kws={'label': 'Percentage (%)'})
ax.set_title('Urgency Level by Category', fontsize=14, fontweight='bold')
ax.set_xlabel('Urgency Level', fontsize=12)
ax.set_ylabel('Category', fontsize=12)
plt.tight_layout()
plt.savefig('05_urgency_heatmap.png', dpi=300, bbox_inches='tight')
plt.show()
print("\n✓ Saved: 05_urgency_heatmap.png")

# ================================
# 11. AGE GROUP ANALYSIS
# ================================

print("\n" + "="*60)
print("AGE GROUP ANALYSIS")
print("="*60)

age_counts = df['Age_Category'].value_counts()
print("\nKnowledge Contributions by Age Category:")
for age, count in age_counts.items():
    print(f"  {age}: {count} contributions ({count/len(df)*100:.1f}%)")

# Age category by term type
age_term_matrix = pd.crosstab(df['Age_Category'], df['Category'], normalize='columns') * 100

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Bar plot
age_counts.plot(kind='bar', ax=axes[0], color='coral', edgecolor='black')
axes[0].set_title('Knowledge Contributions by Age Group', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Age Category', fontsize=12)
axes[0].set_ylabel('Number of Contributions', fontsize=12)
axes[0].tick_params(axis='x', rotation=0)

# Heatmap
sns.heatmap(age_term_matrix, annot=True, fmt='.1f', cmap='Greens', ax=axes[1])
axes[1].set_title('Age Category by Term Category (%)', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Category', fontsize=12)
axes[1].set_ylabel('Age Category', fontsize=12)
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.savefig('06_age_analysis.png', dpi=300, bbox_inches='tight')
plt.show()
print("\n✓ Saved: 06_age_analysis.png")

# ================================
# 12. MAA TERMS DICTIONARY
# ================================

print("\n" + "="*60)
print("MAA TERMS DICTIONARY")
print("="*60)

maa_dict = df[df['Maa_Term'] != 'Not specified'][['Maa_Term', 'Term', 'Definition']].drop_duplicates(subset=['Maa_Term'])
print(f"\nFound {len(maa_dict)} unique Maa terms:")

for idx, row in maa_dict.iterrows():
    print(f"\n📖 {row['Maa_Term']}")
    print(f"   → English: {row['Term']}")
    print(f"   → Definition: {row['Definition'][:100]}...")

# Save Maa dictionary
maa_dict.to_csv("maa_terms_dictionary.csv", index=False, encoding='utf-8')
print("\n✓ Saved: maa_terms_dictionary.csv")

# ================================
# 13. SOURCE/BOMA ANALYSIS
# ================================

print("\n" + "="*60)
print("SOURCE/BOMA ANALYSIS")
print("="*60)

source_counts = df['Source'].value_counts()
print("\nContributions by Meeting Location/Source:")
for source, count in source_counts.items():
    print(f"  {source}: {count} contributions")

fig, ax = plt.subplots(figsize=(10, 6))
source_counts.plot(kind='bar', ax=ax, color='lightgreen', edgecolor='black')
ax.set_title('Knowledge Contributions by Meeting Location', fontsize=14, fontweight='bold')
ax.set_xlabel('Source/Boma', fontsize=12)
ax.set_ylabel('Number of Contributions', fontsize=12)
ax.tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.savefig('07_source_analysis.png', dpi=300, bbox_inches='tight')
plt.show()
print("\n✓ Saved: 07_source_analysis.png")

# ================================
# 14. CAUSE-EFFECT NETWORK ANALYSIS
# ================================

print("\n" + "="*60)
print("CAUSE-EFFECT RELATIONSHIP ANALYSIS")
print("="*60)

# Extract top causes and effects
all_causes_list = df['Cause'].dropna().tolist()
all_effects_list = df['Effect'].dropna().tolist()

# Count cause categories
cause_categories = {
    'Climate-related': ['lack of rain', 'drought', 'climate change', 'dry period', 'no rainfall'],
    'Human-related': ['education', 'subdivision', 'overgrazing', 'deforestation', 'population', 'broken pipes'],
    'Environmental': ['bare land', 'salinity', 'grass scarcity', 'wind', 'erosion']
}

cause_counts = {cat: 0 for cat in cause_categories}
for cause in all_causes_list:
    cause_lower = str(cause).lower()
    for cat, keywords in cause_categories.items():
        if any(kw in cause_lower for kw in keywords):
            cause_counts[cat] += 1
            break

print("\nCauses Grouped by Type:")
for cat, count in cause_counts.items():
    print(f"  {cat}: {count} occurrences")

# Effect categories
effect_categories = {
    'Death/Loss': ['death', 'die', 'dead', 'kill', 'loss'],
    'Migration/Displacement': ['migrate', 'move', 'displacement', 'movement'],
    'Resource Scarcity': ['no grass', 'no milk', 'scarcity', 'shortage', 'hunger'],
    'Disease/Health': ['disease', 'sick', 'measles', 'health']
}

effect_counts = {cat: 0 for cat in effect_categories}
for effect in all_effects_list:
    effect_lower = str(effect).lower()
    for cat, keywords in effect_categories.items():
        if any(kw in effect_lower for kw in keywords):
            effect_counts[cat] += 1
            break

print("\nEffect Types:")
for cat, count in effect_counts.items():
    print(f"  {cat}: {count} occurrences")

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Causes pie
axes[0].pie(cause_counts.values(), labels=cause_counts.keys(), autopct='%1.1f%%', colors=['lightcoral', 'lightblue', 'lightgreen'])
axes[0].set_title('Causes by Type', fontsize=14, fontweight='bold')

# Effects pie
axes[1].pie(effect_counts.values(), labels=effect_counts.keys(), autopct='%1.1f%%', colors=['salmon', 'gold', 'lightgray', 'plum'])
axes[1].set_title('Effects by Type', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.savefig('08_cause_effect_analysis.png', dpi=300, bbox_inches='tight')
plt.show()
print("\n✓ Saved: 08_cause_effect_analysis.png")

# ================================
# 15. SENTIMENT ANALYSIS
# ================================

print("\n" + "="*60)
print("SENTIMENT ANALYSIS")
print("="*60)

def get_sentiment(text):
    if pd.isna(text):
        return 0
    blob = TextBlob(str(text))
    return blob.sentiment.polarity

df['Sentiment'] = df['Definition'].apply(get_sentiment)
avg_sentiment = df['Sentiment'].mean()
sentiment_by_category = df.groupby('Category')['Sentiment'].mean().sort_values()

print(f"\nAverage Sentiment Score: {avg_sentiment:.3f}")
print("(Negative = concerning/urgent, Positive = hopeful/neutral)")

print("\nSentiment by Category:")
for cat, sentiment in sentiment_by_category.items():
    print(f"  {cat}: {sentiment:.3f}")

fig, ax = plt.subplots(figsize=(10, 6))
sentiment_by_category.plot(kind='barh', ax=ax, color='steelblue', edgecolor='black')
ax.set_title('Sentiment Analysis by Category', fontsize=14, fontweight='bold')
ax.set_xlabel('Average Sentiment Score', fontsize=12)
ax.set_ylabel('Category', fontsize=12)
ax.axvline(x=0, color='red', linestyle='--', linewidth=2, label='Neutral')
ax.legend()
plt.tight_layout()
plt.savefig('09_sentiment_analysis.png', dpi=300, bbox_inches='tight')
plt.show()
print("\n✓ Saved: 09_sentiment_analysis.png")

# ================================
# 16. EXPORT ALL DATA
# ================================

print("\n" + "="*60)
print("EXPORTING DATA")
print("="*60)

# Export main dataset
df.to_csv("kimaasai_complete_analysis_dataset.csv", index=False, encoding='utf-8')
print("✓ Saved: kimaasai_complete_analysis_dataset.csv")

# Export summary statistics
summary_stats = df.groupby('Term').agg({
    'Urgency': lambda x: x.mode()[0] if len(x) > 0 else 'Unknown',
    'Category': lambda x: x.mode()[0] if len(x) > 0 else 'Unknown',
    'Definition': 'count'
}).rename(columns={'Definition': 'Mentions'})
summary_stats.to_csv("term_summary_statistics.csv", encoding='utf-8')
print("✓ Saved: term_summary_statistics.csv")

# Export analysis report
with open("analysis_report.txt", "w", encoding='utf-8') as f:
    f.write("="*60 + "\n")
    f.write("KIMASAI ENVIRONMENTAL TERMS ANALYSIS REPORT\n")
    f.write("="*60 + "\n\n")

    f.write(f"Date of Analysis: March 2026\n")
    f.write(f"Source Document: WORD DEFINATION OF TERMS.docx\n\n")

    f.write("DATASET SUMMARY:\n")
    f.write(f"  • Total entries analyzed: {len(df)}\n")
    f.write(f"  • Unique terms: {df['Term'].nunique()}\n")
    f.write(f"  • Unique Maa terms: {df['Maa_Term'].nunique()}\n")
    f.write(f"  • Categories represented: {df['Category'].nunique()}\n\n")

    f.write("TOP 10 MOST FREQUENT TERMS:\n")
    for term, count in term_counts.head(10).items():
        f.write(f"  {term}: {count} mentions\n")

    f.write("\nCATEGORY DISTRIBUTION:\n")
    for cat, count in category_counts.items():
        f.write(f"  {cat}: {count} ({count/len(df)*100:.1f}%)\n")

    f.write("\nURGENCY DISTRIBUTION:\n")
    for urgency, count in df['Urgency'].value_counts().items():
        f.write(f"  {urgency}: {count} ({count/len(df)*100:.1f}%)\n")

    f.write("\nAGE GROUP CONTRIBUTIONS:\n")
    for age, count in age_counts.items():
        f.write(f"  {age}: {count} ({count/len(df)*100:.1f}%)\n")

    f.write("\nMAA TERMS IDENTIFIED:\n")
    for maa in maa_dict['Maa_Term'].tolist():
        f.write(f"  • {maa}\n")

    f.write("\n" + "="*60 + "\n")
    f.write("KEY INSIGHTS\n")
    f.write("="*60 + "\n\n")

    f.write("1. MULTIPLE DROUGHT CONCEPTS:\n")
    f.write("   The community distinguishes several types of drought:\n")
    f.write("   - Olameyu (seasonal drought without death)\n")
    f.write("   - Enkitoto (severe drought with animal death)\n")
    f.write("   - Esoim-osi (drought causing no milk)\n\n")

    f.write("2. CAUSES ARE BOTH CLIMATIC AND HUMAN:\n")
    f.write("   While lack of rain is primary, education, land subdivision,\n")
    f.write("   and population growth are equally emphasized as drivers.\n\n")

    f.write("3. TRADITIONAL KNOWLEDGE PERSISTS:\n")
    f.write("   Elders (50+ years) contribute star-based rangeland monitoring\n")
    f.write("   and migration guidance using natural indicators.\n\n")

    f.write("4. URGENCY IS HIGHEST FOR CLIMATE DISASTERS:\n")
    f.write("   Terms like prolonged drought and severe drought show\n")
    f.write("   highest urgency levels with mentions of death and displacement.\n\n")

    f.write("5. CULTURAL VALUES REMAIN STRONG:\n")
    f.write("   Eramatare (stewardship) encompasses care for livestock,\n")
    f.write("   family, land, and wildlife - showing holistic worldview.\n")

print("\n✓ Saved: analysis_report.txt")

# ================================
# 17. FINAL SUMMARY
# ================================

print("\n" + "="*60)
print("ANALYSIS COMPLETE - SUMMARY")
print("="*60)

print(f"""
📊 Analysis Summary:
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
  Total entries analyzed:     {len(df)}
  Unique terms:               {df['Term'].nunique()}
  Unique Maa terms:           {df['Maa_Term'].nunique()}
  Categories represented:     {df['Category'].nunique()}
  Age groups represented:     {df['Age_Category'].nunique()}
  Sources/Bomas:              {df['Source'].nunique()}

📈 Key Findings:
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
  • Most discussed term:      {term_counts.index[0]} ({term_counts.iloc[0]} mentions)
  • Most common category:     {category_counts.index[0]} ({category_counts.iloc[0]} entries)
  • Primary urgency level:    {df['Urgency'].mode()[0]}
  • Most knowledgeable age:   {age_counts.index[0]} ({age_counts.iloc[0]} contributions)
  • Average sentiment:        {avg_sentiment:.3f}

📁 Output Files Generated:
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
  1. 01_term_frequency.png              - Term frequency visualization
  2. 02_category_analysis.png           - Category distribution
  3. 03_wordclouds.png                  - Word clouds (definitions, causes, effects)
  4. 04_effect_analysis.png             - Effect descriptors
  5. 05_urgency_heatmap.png             - Urgency by category
  6. 06_age_analysis.png                - Age group contributions
  7. 07_source_analysis.png             - Source/Boma analysis
  8. 08_cause_effect_analysis.png       - Cause-effect relationships
  9. 09_sentiment_analysis.png          - Sentiment by category
  10. kimaasai_complete_analysis_dataset.csv - Full dataset
  11. maa_terms_dictionary.csv          - Maa-English dictionary
  12. term_summary_statistics.csv       - Summary by term
  13. analysis_report.txt               - Complete text report

✅ Analysis complete! All files saved in current directory.
""")

print("\n" + "="*60)
print("END OF ANALYSIS")
print("="*60)
```
